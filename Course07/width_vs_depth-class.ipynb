## Width vs Depth

Our goal here is to compare the performances of basic networks.
We will create both very wide and very deep networks, and see which ones are better.

We will try to fit a sequence of functions with increasing complexity, both with a wide and with a deep network.
The first part concentrates on the minimal amount of neurons needed to fit the function with an optimal network (setting the weights and biases manually).
Then, the second part studies the training of the same networks, to fit the same functions. 

### Theory

Taking the function of interest: $f: \mathbb{R} \to \mathbb{R}$ to be linear by segment, with 4 segments:
- $f(x) = 0$ on $\left] -\infty, 0 \right]$
- $f(x) = 2x$ on $\left] 0, \frac{1}{2} \right]$
- $f(x) = 2-2x$ on $\left] \frac{1}{2}, 1 \right]$
- $f(x) = 0$ on $\left] 1, \infty \right]$

Define $f : x\: \rightarrow f(x)$ as a python function, using numpy.

Let also:
- $g(x, 2) = f \circ f(x)$
- $g(x, 3) = f \circ f \circ f(x)$
- $g(x, 4) = f \circ f \circ f \circ f(x)$
- etc...

Define $g : x\: \rightarrow g(x, l)$ as a python function for all $l \in \mathbb{N}^*$.

Plot $f$ and $g$ on $\left] -0.2, 1.2 \right]$

```python
from cProfile import label


def f(x):
    if x <= 0.0:
        return 0.0
    elif x <= 0.5:
        return 2*x
    elif x <= 1.0:
        return 2-2*x
    else:
        return 0.0

def f(x):
    return (1-2*np.abs((x-0.5))).clip(0, None)

def g(x, l):
    for _ in range(l):
        x = f(x)
    return x

x = np.linspace(-1, 2, 1001)
plt.plot(x, f(x), label='f(x)')
for l in range(4):
    plt.plot(x, g(x,l), label=f'g(x,{l})')
plt.legend()
```

### Residual connections

We saw that deep networks are much more efficient than wide networks in theory, but very hard to train in practice.

Residual connections (He et al., 2015, *ResNet*) solve this problem: instead of replacing its input, each hidden layer *adds a correction* to it:
- plain hidden layer: $h_{k+1} = \text{ReLU}(W_k h_k + b_k)$
- residual hidden layer: $h_{k+1} = h_k + \text{ReLU}(W_k h_k + b_k)$

Define a `ResNet` class, with the same parameters as `Net`, but with residual hidden layers (keep `fc_in` followed by a ReLU, and `fc_out`, as in `Net`).
Initialize the weights and biases of all the residual hidden layers to $0$ (see `nn.init.zeros_`).

```python
class ResNet(torch.nn.Module):
    def __init__(self, n, l):
        super().__init__()
        self.fc_in = torch.nn.Linear(1, n)
        self.fc_out = torch.nn.Linear(n, 1)
        self.fc_middle = []
        for _ in range(l-1):
            fc = torch.nn.Linear(n, n)

            # nn.init.zeros_(fc.weight)
            # nn.init.zeros_(fc.bias)
            self.fc_middle.append(fc)
        self.fc_middle = nn.ModuleList(self.fc_middle)
    def forward(self, x):
        x = self.fc_in(x)
        # x = F.relu(x)
        for fc in self.fc_middle:
            # No residual:
            # x = F.relu(fc(x))
            # Residual with gradient stuck:
            # x = x + F.relu(fc(x))
            # Working residual
            x = x + fc(F.relu(x))
        x = F.relu(x)
        x = self.fc_out(x)
        return x

net = ResNet(3, 4)
net(torch.tensor([5], dtype=torch.float32))
```

With this initialization, what function does a `ResNet` compute before training? Why is it a good starting point?

Each residual hidden layer is the identity at initialization: the `ResNet` starts as a network with a single hidden layer, which is easy to train.
During training, the hidden layers progressively learn corrections, i.e. add depth to the network.
Moreover, the gradient can flow from the output to the first layers directly through the identity connections, so it cannot vanish, however deep the network is.

Train a `ResNet` with the same size as the deep network above (50 hidden layers of 15 neurons), with the same learning rate and about $5000$ epochs.

```python
level = 4
n_theory = 3
net = ResNet(3*n_theory, 3*level)
plt.plot(x, g(x, level))
plt.plot(x, net(x).detach())
plt.show()

opt = torch.optim.Adam(net.parameters(), lr=0.001)
x = torch.linspace(-0.1, 1.1, 1000).view(-1, 1)
loss_func = torch.nn.MSELoss()
losses = []
for epoch in range(15_000):
    y_pred = net(x)
    y_true = g(x, level)
    loss = loss_func(y_pred, y_true)
    loss.backward()
    opt.step()
    opt.zero_grad()
    losses.append(loss.item())
plt.plot(x, g(x, level))
plt.plot(x, net(x).detach())
plt.show()
plt.semilogy(losses)
```

*Despite its 50 layers, the residual network converges and fits $g$ closely; it even reaches a much lower loss than the wide network, with 3 times fewer epochs (the spikes in the loss are a usual behaviour of Adam).*

Residual connections let the gradient flow through the identity connections, which makes very deep networks trainable.

*This is why residual connections are found in almost all modern deep architectures (ResNets for images, Transformers for text, ...).*
